###What is the total downtime for each Shift or monthe or Machine?

```python
              # Sum For Shift
Total_downtime_per_Shift = df.groupby('Shift')['Duration_hours'].sum()
print(Total_downtime_per_Shift)
```

```python
          # count For Shift
count_downtime_per_Shift = df.groupby('Shift')['Duration_hours'].count()
print(count_downtime_per_Shift)
```

```python
            # Pie chart for downtime and count shift

plt.figure(figsize=(10, 6))
plt.pie(
    Total_downtime_per_Shift,
    labels=Total_downtime_per_Shift.index,
    autopct='%1.1f%%'

)

# Create legend labels with values
legend_labels = [f"{label}: {value}" for label, value in zip(count_downtime_per_Shift.index,
                                                             count_downtime_per_Shift.values)]

plt.legend(legend_labels, title="Shift Counts", loc="best")

plt.title('Total and Count Downtime by Shift ')
plt.show()
```

Shift B has slightly higher downtime per event, suggesting either more complex operations, higher failure rates, or less efficient recovery.

```python
      # Duration per downtime for each shift
Total_Runtime_per_Shift = df.groupby('Shift')['Runtime_Hours_Day'].sum()
print(Total_Runtime_per_Shift)
```

```python

plt.figure(figsize=(12, 6))
plt.pie(Total_Runtime_per_Shift,labels=Total_Runtime_per_Shift.index,autopct='%1.1f%%')
plt.title('Total Runtime by Shift' , fontsize = 20)
plt.show()
```

Shift A has the highest runtime, but combined with downtime data, Shift B appears to be the least efficient overall.

```python
import warnings
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
plt.style.use('dark_background')
import plotly.graph_objects as go
warnings.filterwarnings("ignore")
```

```python
file_path = '/content/Manufacturing_Downtime (1).xlsx'

all_sheets_data = pd.read_excel(file_path, sheet_name=None)

print(list(all_sheets_data.keys()))
```

```python
Machine_Downtime = all_sheets_data['Manufacturing_Downtime']
Machine_Downtime.head(5)
```

```python
Machine_Info_df = all_sheets_data['Machine_Info']
Machine_Info_df.head(5)
```

```python
def rating(Machine_Downtime) :
  if Machine_Downtime['Efficiency_Score']<=.5:
    return 'Poor'
  elif Machine_Downtime['Efficiency_Score']<=.9:
    return 'Good'
  elif Machine_Downtime['Efficiency_Score']<=1:
    return'Excellent'
  elif Machine_Downtime['Efficiency_Score']>1:
    return'Amazing'
Machine_Downtime['Efficiency_status']=Machine_Downtime.apply(rating,axis=1)

Machine_Downtime
```

```python
def rating(Machine_Downtime) :
  if Machine_Downtime['Efficiency_Score']<=.5:
    return 'Poor'
  elif Machine_Downtime['Efficiency_Score']<=.9:
    return 'Good'
  elif Machine_Downtime['Efficiency_Score']<=1:
    return'Excellent'
  elif Machine_Downtime['Efficiency_Score']>1:
    return'Amazing'
Machine_Downtime['Efficiency_status']=Machine_Downtime.apply(rating,axis=1)

Machine_Downtime
```

```python
Machine_target_actual=Machine_Info_df.groupby('Machine_ID').agg({'Total Runtime_Hours_year':'sum','Target_Runtime_Hours_year':'sum'}).sort_values(by='Target_Runtime_Hours_year',ascending=False)
Machine_target_actual
```

```python
# 2. إعداد الرسم البياني
fig, axes = plt.subplots(figsize=(50, 25))
# 3. رسم الأعمدة لساعات العمل الفعلية
axes.bar(
    Machine_target_actual.index,             # على محور X: أسماء الماكينات
    Machine_target_actual['Total Runtime_Hours_year'], # على محور Y: الساعات الفعلية
    color='skyblue',          # لون الأعمدة
    label='Runtime_Hours_year'
)

# 4. رسم الخط لساعات الهدف (التارجت)
axes.plot(
    Machine_target_actual.index,            # على محور X: أسماء الماكينات
    Machine_target_actual['Target_Runtime_Hours_year'],         # على محور Y: ساعات الهدف
    color='red',              # لون الخط
    linestyle='--',           # نوع الخط (متقطع)
    marker='o',               # شكل النقاط على الخط
    label='Target',
    linewidth=2               # سمك الخط
)

# 5. إضافة التفاصيل والتنسيقات (تنسيق عربي)
axes.set_xlabel('Machine_ID', fontsize=12)
axes.set_ylabel('Hours_year', fontsize=12)
axes.set_title('Target_by_actual', fontsize=14)
axes.legend() # عرض مفتاح الرسم البياني

# 7. عرض الرسم البياني
plt.tight_layout()
plt.show()
```
